## Meanders in Manim

A meander is defined by its upper and lower blocks. `MeanderElement` constructs the nodes and arcs as well as automatically framing the camera.

A little explainer on the `%%manim ...` portion in the first line of the following cells:

The `%%manim ...` line is *required* to be present, and *required* to be the first line in a cell that will output a Manim animation in a jupyter notebook such as this.

The `-qm` flag changes the quality of the animation, currently set to 'medium' quality.
You can change it to `-q [l|m|h|p|k]` for 'low', 'medium', 'high', 'pro', and '4k' respectively, but keep in mind the higher resolution you chooser, the longer it takes to render.

The `-v warning` flag changes what Manim prints out, this means that general debug or miscellaneous information won't be printed and clutter up the output, only anything classified as a warning or something more severe will be shown.

The `--disable_caching` flag isn't necessary, but it speeds up the rendering for our use-case. The meanders consist of many moving elements which update on every frame, so by telling Manim not to cache them it speeds up rendering.

Lastly the name of the actual scene must be given so that Manim knows what to render, in this case `VisualizingMeander` is given.

```python
%%manim -qm -v warning --disable_caching VisualizingMeander

class VisualizingMeander(MovingCameraScene):
    def construct(self):
        m = Meander([8, 5], [5, 6, 2])
        m_elem = MeanderElement(m)
        self.add(m_elem)
        m_elem.fit_camera(self)
        self.wait(1)
```

## Winding Moves

### Rotation Contraction ($b_1 < a_1 < 2b_1$)

```python
%%manim -qm -v warning --disable_caching RotationContractionExample

class RotationContractionExample(MovingCameraScene):
    def construct(self):
        m = Meander([8, 5], [5, 6, 2])
        m_elem = MeanderElement(m)
        self.add(m_elem)
        m_elem.fit_camera(self)
        self.wait(1)
        m_elem.rotation_contraction_animation(self)
```

```python
%%manim -qm -v warning --disable_caching PureContractionExample

class PureContractionExample(MovingCameraScene):
    def construct(self):
        m = Meander([8, 5], [3, 6, 4])
        m_elem = MeanderElement(m)
        self.add(m_elem)
        m_elem.fit_camera(self)
        self.wait(1)
        m_elem.pure_contraction_animation(self)
```

## Automatic Full Wind-Down

`wind_down_animation()` automatically evaluates the next move at each step and animates the full reduction until the meander is the empty meander.

```python
%%manim -qm -v warning --disable_caching FullWindDownExample

class FullWindDownExample(MovingCameraScene):
    def construct(self):
        m = Meander([8, 2], [2, 6, 2])
        m_elem = MeanderElement(m)
        self.add(m_elem)
        m_elem.fit_camera(self)
        self.wait(1)
        m_elem.wind_down_animation(self)
        self.wait(1)
```
