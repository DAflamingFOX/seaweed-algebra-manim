# Meander animations with Manim

This notebook provides an interactive walkthrough for animating meander winding moves from seaweed algebras using [Manim](https://www.manim.community/) and [`seaweed-algebra`](https://github.com/daflamingfox/seaweed_algebra).

```python
from manim import *
from seaweed_algebra import Meander

from seaweed_algebra_manim import MeanderElement

# Enable inline video embedding in notebook output
config.media_embed = True
```

## Meanders in Manim

A meander is defined by its upper and lower blocks. `MeanderElement` constructs the nodes and arcs as well as automatically framing the camera.

```python
%%manim -qm -v WARNING VisualizingMeander

class VisualizingMeander(MovingCameraScene):
    def construct(self):
        m = Meander([8, 5], [5, 6, 2])
        m_elem = MeanderElement(m)
        self.add(m_elem)
        m_elem.fit_camera(self)
        self.wait(1)
```

## Winding Moves

### Rotation Contraction ($b_1 < a_1 < 2b_1$)

```python
%%manim -qm -v WARNING RotationContractionExample

class RotationContractionExample(MovingCameraScene):
    def construct(self):
        m = Meander([8, 5], [5, 6, 2])
        m_elem = MeanderElement(m)
        self.add(m_elem)
        m_elem.fit_camera(self)
        self.wait(1)
        m_elem.rotation_contraction_animation(self)
```

## Automatic Full Wind-Down

`wind_down_animation()` automatically evaluates the next move at each step and animates the full reduction until the meander is the empty meander.

```python
%%manim -qm -v WARNING FullWindDownExample

class FullWindDownExample(MovingCameraScene):
    def construct(self):
        m = Meander([8, 2], [2, 6, 2])
        m_elem = MeanderElement(m)
        self.add(m_elem)
        m_elem.fit_camera(self)
        self.wait(1)
        m_elem.wind_down_animation(self)
        self.wait(1)
```
